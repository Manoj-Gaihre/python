# Comprehensions
 one line instead of a loop

 ### list comprehensions


```python
# old way
num = []
for i in range(1,6):
  num.append(i**2)

# comprehension way
num = [i**2 for i in range(1,6)]

# with a filter
evens = [i for i in num if i%2 == 0]
print(evens)
```

```python
# dict and set comprehensions

# dict comprehension
squares = {i: i**2 for i in range(1,6)}

# set comprehension
nums = [1,2,2,3,4,4]
evens = {x for x in nums if x % 2 == 0}
```

# functional programming
lambda, map and filter

```python
nums = [1,2,3,4,5]

square = lambda x: x * x
squares = list(map(square, nums))

evens = list(filter(lambda x:x % 2 == 0, nums))
```

```python
def square(x):
  return x*x

squared = square(5)
print(squared)
```

```python
square_value = lambda x:x*x
square_value(3)
```

```python
# map
results = list(map(square,nums))
print(results)
```

```python
# filter
even_results = list(filter(lambda x:x % 2 == 0, nums))
print(even_results)
```

# reduce and sorted(key= ...)

```python
# reduce() folds a list down to one value, step by step
from functools import reduce
nums = [1,2,3,4]
total = reduce(lambda a, b: a+b, nums)
print(total)
```

```python
people = [
    {'name': 'ram', 'age': 30},
    {'name': 'sam', 'age': 25}
]
print(sorted(people, key =lambda p:p['age'], reverse = True))
print(sorted(people, key =lambda p:p['age']))
```

# decorators
a function inside another function.

```python
from logging import log
def long(func):
  def wrapper(*args):
    print(f'calling{func.__name__}')
    return func(*args)
  return wrapper

@log
def greet(name):
  return f'hello {name}'

greet('ram')
```

```python
def log(func):
    def wrapper(*args):
        print("Transaction started")
        result = func(*args)
        print("Transaction completed")
        return result

    return wrapper


@log
def withdraw(name, amount):
    return f"{name} withdrew Rs. {amount}"


print(withdraw("Ram", 5000))
```

```python
def log(func):
    def wrapper(*args):
        print(f'Calling {func.__name__}')
        return func(*args)
    return wrapper

@log
def greet(name):
    return f'hello {name}'

greet('Ram')
```

# properties

```python
class Person:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        self._name = value.title()

p = Person('ram')
p.name = 'shyam'
print(p.name)

class Person:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        self._name = value.title()

p = Person('ram')
p.name = 'shyam'
print(p.name)   # Shyam
```
