```python
flat = np.arange(1,22)
```

```python
flat.shape
```

## Views vs. copies: what slicing actually returns

```python
first_three = temps[0:3]
first_three[0] = 100
print(temps) # changed too
```

```python
temps = np.array([22,24,18,21,23,25,26])
```

```python
# for data manipulation do copy
safe_copy = temps[0:3].copy()  # now independent
```

```python
safe_copy[0] = 100
```

```python
print(temps)
print(safe_copy)
```

## vectorized operation

```python
# The loop way 
fahrenheit = []
for c in temps:
    fahrenheit.append(c * 9/5 + 32)
```

```python
# the vectorized way
fahrenheit = c * 9/5 + 32
# it does element - wise arithmetic
```

## Aggregating along an axis
axis = 0 walks down the rows(collapses cities, keeps 7 days in grid)
axis = 1 walks acorss the columns(collapses days, keeps 3 cities)

```python
grid.sum(axis = 0)
```

```python
grid.sum(axis = 1)
```

```python
# daywise min temperature
grid.min(axis = 0)
```

```python
grid.min(axis = 1)
```

```python
# get the position of maximum temperature across city for each day
print(grid.argmax(axis = 0))
```
