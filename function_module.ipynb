# default argument
If we donot pass any value while calling the function, then using the value that passed during the definition of the function is default argument.

```python
def greet(name, greet_msg="hi"):
  print(f"{greet_msg}, {name}")

greet("manoj")
greet("manoj","hello")
```

### write a function to calculate total amount:

input: quantity, price, discount = 0%

```python
def calculate_total(quantity:int, price:float, discount:float = 0):
  total_amount = (quantity * price)
  discount_amount = (discount/100)*total_amount
  total_amount = total_amount-discount_amount
  print(total_amount)

calculate_total(10,5) # order argument
calculate_total(10,5,5)
calculate_total(quantity=5, discount=5, price=10) # keyword argument
```

# *args and **kwargs

```python
def add_all(*args):
  print(type(args))
  print(args)
  print(sum(args))

add_all(2,3,4,5) # function number argument -> unknown in advance # ordered argument
```

```python
# **kwargs   # accept keyword argument
def print_profile(**kwargs):
  print(type(kwargs))
  print(kwargs)
  print("\n")
  for key, value in kwargs.items():
    print(f"{key} - {value}")

```

```python
student = {
    "name": "hari",
    "address":"pokhara",
    "age": 21
}
```

```python
print_profile(name="hari", address="pokhara", age=21)
```

# built in function

```python
# map(),
# filter()
# zip()
# enumerate()
```

```python
# map
# map() is used when you want to apply the same function to every item in an iterable such as a list, tuple, etc.
def is_even(n):
  return n%2==0
```

```python
numbers = [2,3,4,5,6]

# map(function, iterable)
list(map(is_even, numbers))
```

```python
#  if done using lambda
list(map(lambda n:n%2 ==0, numbers))
```

# filter
### filter() is used when you want to select only the elements that satisfy a condition but map() iterates through all the element including True and False.

```python
list(filter(lambda n:n%2 ==0, numbers))
```

# Virtual Environment

The main reason we use it is to keep the packages and dependencies of one project separate from another project.

│

└── venv

└── venv
