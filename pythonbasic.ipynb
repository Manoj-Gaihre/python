**Is Python a Compiler or an Interpreter?**

The answer is:

Python is an interpreted language. Python first compiles source code into bytecode, then the Python Virtual Machine interprets and executes that bytecode.

**REPL in Python**

REPL stands for:

R – Read
E – Evaluate
P – Print
L – Loop

REPL is Python's interactive interpreter, where you can type Python code one line at a time and get the result immediately.

How REPL Works

1.   Read – Reads the Python statement you enter.
2.   Evaluate – Executes the statement.


1.   Print – Displays the result.
2.   Loop – Waits for the next statement.








**Scripting Mode vs Interactive Mode in Python**

Python can be used in two modes:
Interactive Mode (REPL)
Scripting Mode



1.   Interactive Mode
Interactive mode allows you to write and execute Python code one line (one statement) at a time. The result is displayed immediately. There is no need to save the code and it is best for learning, testing, and debugging.
2.   Scripting Mode         
In scripting mode, you write your code in a Python file (.py), save it, and then run the entire program. It is suitable for medium and large applications.

**Real-Life Analogy**

Imagine you are solving math problems.

Interactive Mode: You solve one question on a whiteboard. After finishing, you erase it and solve the next one.
Scripting Mode: You write all the questions and solutions in a notebook. You can keep the notebook, review it later, and share it with others.

```python
#  lists
regions = ["east", "west", "north", "south"]

print(regions[0])
print(regions[1])
print(regions[2])
print(regions[3])

print(regions[0:2])

# list append---> add a single new items in the last of the given list
regions.append("NW")
print(regions)

#  list extend ---> adds multiple items at the end of the given list and should be given as the list
regions.extend(["NE","SW","SE"])
print(regions)

# list insert ---> add the single value in the specified value
regions.insert(0, "new region")
print(regions)

# list remove ----> removes the first matching value from the list
regions.remove("new region")
print(regions)

regions.extend(["NE","SW","SE"])
print(regions)

regions.remove("NE")
print(regions)

# pop list ---> returns the removed value and usually remove the last value
popped_region = regions.pop()
print("popped region: ", popped_region)
print(regions)

popped_region = regions.pop(3)
print("popped region: ", popped_region)
print(regions)

# sort
regions.sort()
print(regions)

regions.sort(reverse=True)
print(regions)
```

**Why dictionaries**

We use dictionaries when we want to store and retrieve data using keys instead of indexes. They are ideal for representing data that has a clear relationship between a key and its value.

```python
# converting dictionaries to list or tuple or sets

# list
student = {
    "name": "Manoj",
    "age": 22,
    "country": "Nepal"
}
# Converting keys to list
keys = list(student)
print(keys)

# Converting values to list
values = list(student.values())
print(values)

# Converting key values pairs to list
items = list(student.items())
print(items)


# converting to tuples
student = {
    "name": "Manoj",
    "age": 22,
    "country": "Nepal"
}

keys = tuple(student)
print(keys)

values = tuple(student.values())
print(values)

items = tuple(student.items())
print(items)


# Converting to sets
student = {
    "name": "Manoj",
    "age": 22,
    "country": "Nepal"
}

keys = set(student)
print(keys)

values = set(student.values())
print(values)

items = set(student.items())
print(items)
```

```python
# How would you write a function or loop to update a specific inner key across all parent dictionaries, regardless of how many outer keys exist?
students = {
    "ram":{"age":20, "status":"active"},
    "sita":{"age":19, "status":"inactive"},
    "hari":{"age":21, "status":"active"}
}

for inner_dict in students.values():
  inner_dict["age"] += 1
  inner_dict["status"] = "inactive"

print(students)
```
