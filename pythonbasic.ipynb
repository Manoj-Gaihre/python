**Is Python a Compiler or an Interpreter?**

The answer is:

Python is an interpreted language. Python first compiles source code into bytecode, then the Python Virtual Machine interprets and executes that bytecode.

**REPL in Python**

REPL stands for:

R – Read
E – Evaluate
P – Print
L – Loop

REPL is Python's interactive interpreter, where you can type Python code one line at a time and get the result immediately.

How REPL Works

1.   Read – Reads the Python statement you enter.
2.   Evaluate – Executes the statement.


1.   Print – Displays the result.
2.   Loop – Waits for the next statement.








**Scripting Mode vs Interactive Mode in Python**

Python can be used in two modes:
Interactive Mode (REPL)
Scripting Mode



1.   Interactive Mode
Interactive mode allows you to write and execute Python code one line (one statement) at a time. The result is displayed immediately. There is no need to save the code and it is best for learning, testing, and debugging.
2.   Scripting Mode         
In scripting mode, you write your code in a Python file (.py), save it, and then run the entire program. It is suitable for medium and large applications.

**Real-Life Analogy**

Imagine you are solving math problems.

Interactive Mode: You solve one question on a whiteboard. After finishing, you erase it and solve the next one.
Scripting Mode: You write all the questions and solutions in a notebook. You can keep the notebook, review it later, and share it with others.

**Dynamic Typing in Python**

Dynamic typing means that you do not need to declare the data type of a variable. Python automatically determines the type based on the value assigned to the variable.

**Eg**

x = 10

print(type(x))

x = 3.14

print(type(x))

x = "Hello"

print(type(x))

x = True

print(type(x))

**Why is it Called Dynamic?**

The type of the variable is decided while the program is running (runtime), not before execution.

**This is static typing**

int x = 10;

which is done in programming language like C, C++, Java, etc.

**We can have the complex data type as:**

signal = 3 + 4j
