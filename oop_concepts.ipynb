# Classes and Objects

A class is a blueprint; an object is one instance built from it.

__init__ runs automatically when an objcet is created.

self refers to the specific object a method is running on.

```python
class Dog:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def bark(self):
    return f'{self.name} is barking'

d = Dog('Tommy', 3)
d.bark()
```

```python
class Student:
  def __init__(self, name, roll_no, marks = 0):
    self.name = name
    self.roll_no = roll_no
    self.marks = marks

  def study(self):
    return f'{self.name} is studying'

  def get_marks(self):
    return f'{self.name} scored {self.marks} marks.'

  def add_marks(self, points):
    self.marks += points
    return f'{self.name} now has {self.marks} marks'


s = Student('Anita', 12)
print(s.study())
print(s.get_marks())
print(s.add_marks(15))
print(s.get_marks())
```

```python
# type checking
class Phone:
  def __init__(self, brand, price):
    self.brand = brand
    self.price = float(price)

p = Phone("nokia", 1000)
print(p.brand)
print(p.price)
```

# Inheritance

A child class reuses everything from its parent class.

It can override a method to give it its own behavior.

Python looks for a method on the child first, then walks up to the parent.

```python
class Animal:
  def sound(self):
    return 'some sound'

class Dog(Animal):
  def sound(self):
    return 'woof'

print(Dog().sound())
print(Animal().sound())
```

# Encapsulation

A double underscore(__salary) makes an attibute private to the class.

Outside code goes through get/set methods instead of touching it directly.

This lets the class validate data beforeit changes.

```python
class Employee:
  def __init__(self, salary):
    self.__salary = salary

  def get_salary(self):
    return self.__salary

  def set_salary(self, amount):
    if amount > 0:
      self.__salary = amount

e = Employee(500000)
e.set_salary(600000)
e.get_salary()
```

# Polymorphism and Abstraction

Ploymorphis: the same method name behaves differently on each class.

Abstraction: an ABC(abstract class) forces every subclass to implement a method.


```python
from abc import ABC, abstractmethod

class Greet(ABC):
  @abstractmethod
  def say_hello(self): ...

class English(Greet):
  def say_hello(self):
     return "Hello"
class Nepali(Greet):
  def say_hello(self):
    return "Namaste!"

for g in (English(), Nepali()):
  print(g.say_hello())
```
