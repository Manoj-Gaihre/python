## Aggregation data to summarize trends in pandas

```python
df = pd.DataFrame(
    {
        "product":["A","A","B","B","C"],
        "sales":[100,150,200,250,300],
        "quantity":[2,3,4,5,6]
    }
)
```

```python
df
```

```python
df["sales"].sum()
```

```python
df.sum()
```

```python
df.describe()
```

```python
# calculate sales for each product
# group by
df.groupby("product")["sales"].sum()
```

```python
# Multiple aggregations
df.groupby("product")["sales"].agg(
    [
        "sum",
        "mean",
        "min",
        "max"
    ]
)
```

```python
df.groupby("product").agg(
    {
        "sales":["sum","mean","min"],
        "quantity":["sum","mean","min"]
    }
)
```

```python
df.groupby("product").agg(
    total_sales = ("sales","sum"),
    average_sales = ("sales","mean"),
    total_quantity = ("quantity","sum"),
)
```

```python
df.groupby("product").count()
```

```python
df.groupby("product").size()
```

```python
df.groupby("product")["sales"].count()
```

```python
df = pd.DataFrame(
    {
        "date":[
            "2026-01-15",
            "2026-01-15",
            "2026-02-20",
            "2026-02-20",
            "2026-03-10"
        ],
        "sales":[100,150,200,250,300]
    }
)

df["date"] = pd.to_datetime(df["date"])
```

```python
df.groupby("date")["sales"].sum()
```

```python
df["month"] = df["date"].dt.to_period("M")
df.groupby("month")["sales"].sum()
```

```python
df = pd.DataFrame(
    {
        "region": ["East","East","West","West"],
        "product": ["A","B","A","B"],
        "sales": [100,200,150,300]
    }
)

df.groupby(["region","product"])["sales"].sum()
```

```python
pd.pivot_table(
    df,
    values = "sales",
    index = "region",
    columns = "product",
    aggfunc = "sum"
)
```
