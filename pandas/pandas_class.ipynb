### DataFrame

```python
# Create a dataframe from a dictionary: having keys {'name':[], 'score':[]}
score = pd.DataFrame({"name":["Hari","Basanta","Gopi"],
              "score":[22,34,45]})

score
```

```python
type(score)
```

```python
score.index
```

```python
# create a dataframe from list of dictionaries

score = pd.DataFrame([{"name":"hari","score":22},
                      {"name":"basanta","score":34},
                      {"name":"gopi","score":45}])

score
```

```python
# the end like [0,4] here 4 is inclusive in loc
df.loc[0:4,["name","subject","marks"]]
```

```python
# 5 is exclusive in iloc
df.iloc[0:5,1:4]
```

## Boolean filtering

```python
df["marks"] > 50
```

```python
df.head()
```

```python
df[df["marks"]>50]
```

```python
df[(df["marks"]>50) & (df["attendance_percent"] > 70)]
```

```python
# filter students from kathmandu having subjects python programming or machine learning
df[(df["city"]=='Kathmandu') & ((df["subject"] == 'Python Programming') | (df["subject"] == 'Machine Learning'))]
```

## A more readable style: df.query()

```python
df.query("city == 'Kathmandu' and (subject == 'Python Programming' or subject == 'Machine Learning')")
```
